```python
from pathlib import Path
import sys

# Supports opening the notebook from the repository root or notebooks/ in VS Code.
ROOT = Path.cwd()
if not (ROOT / 'portfolio').is_dir():
    ROOT = ROOT.parent
if not (ROOT / 'portfolio').is_dir():
    raise RuntimeError('Open this notebook inside the cloned repository.')
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
DATA = ROOT / 'data'
ARTIFACTS = ROOT / 'artifacts'
ARTIFACTS.mkdir(exist_ok=True)
```

## Index the supplied demonstration corpus

```python
from portfolio.rag import Retriever, Conversation
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

corpus = DATA / 'sample-knowledge-base.txt'
retriever = Retriever.build({corpus.name: corpus.read_text(encoding='utf-8')})
name = 'google/flan-t5-small'
conversation = Conversation(retriever, AutoTokenizer.from_pretrained(name),
                            AutoModelForSeq2SeqLM.from_pretrained(name))
```

## Inspect the answer and retrieved sources

```python
result = conversation.ask('What does the portfolio cover?')
result
```
