## Generate the explicitly synthetic dataset

```python
from portfolio.churn import make_sample
from portfolio.tabular import train
import json

frame = make_sample(n_samples=7043, seed=42)
print('Data origin: synthetic simulator, 7,043 rows, seed 42')
frame.head()
```

## Train, select, and evaluate

```python
model, report, predictions = train(frame, 'Churn', 'classification')
report['data_origin'] = 'synthetic; not the external Telco benchmark'
print(json.dumps(report, indent=2))
predictions.head()
```

## Export a complete preprocessing/model pipeline

```python
import joblib
folder = ARTIFACTS / 'synthetic-churn'
folder.mkdir(parents=True, exist_ok=True)
joblib.dump(model, folder / 'model.joblib')
(folder / 'metrics.json').write_text(json.dumps(report, indent=2))
predictions.to_csv(folder / 'predictions.csv', index=False)
print('Exported to', folder)
```
